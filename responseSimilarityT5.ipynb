```python
import ast

from sqlalchemy import create_engine
from sqlalchemy.orm import sessionmaker
from models.testResult import TestResult
```

```python
engine1 = create_engine('sqlite:///similarity1.db')
Session1 = sessionmaker(bind=engine1)
session1 = Session1()

engine2 = create_engine('sqlite:///similarity2.db')
Session2 = sessionmaker(bind=engine2)
session2 = Session2()

test_results1 = session1.query(TestResult).all()
test_results2 = session2.query(TestResult).all()

result_list1 = []
result_list2 = []

for result in test_results1:
    result_list1.append(ast.literal_eval(result.results_array))
for result in test_results2:
    result_list2.append(ast.literal_eval(result.results_array))
```

```python
print(result_list1)
print(result_list2)
```

```python
def compare_lists_of_lists(input1, input2):
    # Check if the structure of the two inputs is the same
    if len(input1) != len(input2):
        return f"The two inputs have different number of sublists. input 1 has {len(input1)} sublists, input 2 has {len(input2)} sublists."
    
    for sublist1, sublist2 in zip(input1, input2):
        if len(sublist1) != len(sublist2):
            return "The sublists have different lengths."
    
    # Iterate through the lists and compare elements
    differences = []
    for i, (sublist1, sublist2) in enumerate(zip(input1, input2)):
        for j, (elem1, elem2) in enumerate(zip(sublist1, sublist2)):
            if elem1 != elem2:
                differences.append((i, j, elem1, elem2))
    
    if differences:
        return differences
    else:
        return "The two inputs are identical."
```

```python
print(compare_lists_of_lists(result_list1, result_list2))
```
